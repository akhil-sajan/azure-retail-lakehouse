```python
from pyspark.sql import functions as F

dim_product = (
    spark.read.table("databricks_catalog.silver_schema.products")
    .withColumn("product_key", F.xxhash64("product_id"))
    .select("product_key", "product_id", "product_name", "category", "brand",
            "price", "discounted_price", "price_band")
    .withColumn("loaded_at", F.current_timestamp())
)

(dim_product.write.mode("overwrite").option("overwriteSchema", "true")
    .saveAsTable("databricks_catalog.gold_schema.dim_product"))

print("✓ dim_product written:", dim_product.count(), "rows")
```
